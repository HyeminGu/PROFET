# GPA for transportation of gene expression datasets


This is the baseline model for transporting time series of gene expression data.

For each time points, set of gene expression samples form empirical distributions at time $t$, and say $Q_t$.

GPA algorithm implements transportation of one "empirical" distribution to another "empirical" distribution by calculating their $f$-divergence (metric among distributions) 
$$D_f(P\|Q) = E_Q\left[f\left(\frac{dP}{dQ}\right)\right], P \ll Q$$ and form vector fields that is the gradient descent direction of the $f$-divergence. Here are some examples of $f$-divergences:
* KL divergence $D_\text{KL}(P\|Q) = E_P\left[\log \frac{dP}{dQ}\right] = $ with $f(x)=xlogx$
* Reverse KL divergence $D_\text{KL}(Q\|P) = E_Q\left[\log \frac{dQ}{dP}\right]$ with $f(x)=-logx$
* $\alpha$-divergence $D_\alpha(P\|Q)$ with $f(x)=\frac{x^\alpha-1}{\alpha (\alpha-1)}$ for $\alpha > 1$
* Jensen-Shannon divergence with $f(x) = -(x+1) \log \frac{1+x}{2} + x \log x$

For the stability of the algorithm, we use Lipschitz regularized function optimizer $\phi^*$ that optimize the dual formulation of $f$-divergence.

Set $Q_{t_1}$ as a source distribution and $Q_{t_2}$ as a target distribution for $t_1 < t_2$.

The vectorfield to transport particles is given as  $\dot{x} = -\nabla \phi^*(x, t)$ where $\phi^*(x, t)$ is the nonlinear optimizer $\phi^*$ above learned by Machine learning and $t$ is time for GPA algorithm.

### Reduce the dimension by two different ways

1. Select 175 genes from ```EMT_Hallmark_genes.txt``` <br>
There is no *reconstruction map* and therefore we assume that the differences in other genes are neglegent and consider the 175 dimensional space as the original space and the reduced matrix will be stored in ```data/Gene_Expression_Matrix-EMT_Hallmark_genes.txt```. <br>
Then further apply PCA on entire days matrix with dimensions ```d_red = 2, 4, 8, 16, 32, 64, 128, 175 (full)```. <br>
  * PCA projection maps ```pca = decomposition.PCA(n_components=d_red, random_state=0)``` will be stored at ```data/emt_pca_{d_red}.pkl```
  * Projected sample matrix (row: entire days samples, column: PCA vectors in principal order) will be stored at ```data/emt_pca_projected.csv``` along with days information ```data/Cells_Days.txt```



2. Project to PCA basis <br>
PCA provides a *reconstruction map* and therefore we compare the reconstructed GPA-transported dataset with source and target in the original 15K dimensional space. <br>
PCA is applied on entire days matrix with dimensions ```d_red = 2, 4, 8, 16, 32, 64, 128, 175 (for comparison)```. <br>

  * PCA projection maps ```pca = decomposition.PCA(n_components=d_red, random_state=0)``` will be stored at ```data/pca_{d_red}.pkl```
  * Projected sample matrix (row: entire days samples, column: PCA vectors in principal order) will be stored at ```data/pca_projected.csv``` along with days information ```data/Cells_Days.txt```


The point is to **project to a space where the gene expressions on different days are distinguishable**.

Or apply different dimension reduction algorithm in the cell below.

**Other dimensionality reduction methods to explore**
1. based on biological knowledge for validation
2. unbiased dimensionality reduction methods: ex) Force-directed graph drawing, UMAP

````python
# select genes from EMT_Hallmark_genes.txt 
if os.path.exists(data_dir + "Gene_Expression_Matrix-EMT_Hallmark_genes.txt"):
    df_reduced_emt = pd.read_table(data_dir + 'Gene_Expression_Matrix-EMT_Hallmark_genes.txt', sep="\t")
    print(df_reduced_emt)
    col_selection_emt = df_reduced_emt.columns[1:]
elif os.path.exists(data_dir + 'Gene_Expression_Matrix.txt'):
    col_selection_emt = np.loadtxt(data_dir + 'EMT_Hallmark_genes.txt', dtype=str)
    df_reduced_emt = df.filter(items=np.concatenate((['id'], col_selection_emt)))
    print(df_reduced_emt)
    print("%d EMT Hallmark genes are excluded!" % (len(col_selection_emt)-len(df_reduced_emt.columns[1:])))
    col_selection_emt = df_reduced_emt.columns[1:]
    df_reduced_emt.to_csv(data_dir + 'Gene_Expression_Matrix-EMT_Hallmark_genes.txt', sep='\t', index=False)
    
print("Original space dimension = ", len(col_selection_emt))
````

````python
from sklearn import decomposition
````

````python
# Apply PCA to EMT reduced matrix with different dimensions
d_reds = [2, 4, 8, 16, 32, 64, 128, len(col_selection_emt)]

if not os.path.exists(data_dir+"emt_pca_variance_ratios.png"):
    mat_reduced = np.array(df_reduced_emt)[:,1:] # exclude id

    for d_red in d_reds:
        pca = decomposition.PCA(n_components=d_red, random_state=0)
        pca.fit(mat_reduced)
        mat_projected = pca.transform(mat_reduced)
        # mat_reconstructed = pca.inverse_transform(mat_projected)

        # store pca map for each d_red
        pk.dump([pca,], open(data_dir+"emt_pca_"+str(d_red)+".pkl","wb"))

        if d_red == len(col_selection_emt):
            # store projected matrix
            np.savetxt(data_dir+"emt_pca_projected.csv", mat_projected, delimiter=",")

            # plot variance ratio
            variances = pca.explained_variance_ratio_ * 100
            #np.array(pca.explained_variance_)/np.sum(pca.explained_variance_)*100
            plt.plot(range(1, len(variances)+1), np.cumsum(variances), '-')
            plt.xlabel("number of PCA basis", fontsize=16)
            plt.ylabel("Explained variance (%)", fontsize=16)

            for reduced_dim in d_reds[:-1]:
                plt.vlines(x=reduced_dim, ymin=variances[0], ymax=np.sum(variances), color = 'r', linestyles='dashed')
                plt.text(x=reduced_dim+7, y= np.cumsum(variances)[reduced_dim-1]-3, s="$d'=%d$\n%.2f %%" % (reduced_dim, np.cumsum(variances)[reduced_dim-1]), fontsize=16)

            plt.ylim([variances[0], np.sum(variances)])
            plt.tick_params(axis='x', labelsize=14)
            plt.tick_params(axis='y', labelsize=14)
              
            plt.savefig(data_dir+"emt_pca_variance_ratios.png")
            plt.show() 
else:
    image = plt.imread(data_dir+"emt_pca_variance_ratios.png")
    plt.imshow(image)
    plt.axis('off') 
    plt.show()
````

````python
# Project to PCA basis
# d_reds = [2, 4, 8, 16, 32, 64, 128, len(col_selection_emt)]

if os.path.exists(data_dir + 'Gene_Expression_Matrix.txt') and not os.path.exists(data_dir+"pca_variance_ratios.png"):
    mat = np.array(df)[:,1:] # first column is id, not gene 
    for d_red in d_reds:
        pca = decomposition.PCA(n_components=d_red, random_state=0)
        pca.fit(mat)
        mat_projected = pca.transform(mat)
        # mat_reconstructed = pca.inverse_transform(mat_projected)

        # store pca map for each d_red
        pk.dump([pca,], open(data_dir+"pca_"+str(d_red)+".pkl","wb"))

        if d_red == len(col_selection_emt):
            # store projected matrix
            np.savetxt(data_dir+"emt_pca_projected.csv", mat_projected, delimiter=",")

            # plot variance ratio
            variances = pca.explained_variance_ratio_ * 100
            plt.plot(range(1, len(variances)+1), np.cumsum(variances), '-')
            plt.xlabel("number of PCA basis", fontsize=16)
            plt.ylabel("Explained variance (%)", fontsize=16)

            for reduced_dim in d_reds:
                plt.vlines(x=reduced_dim, ymin=variances[0], ymax=np.sum(variances), color = 'r', linestyles='dashed')
                plt.text(x=reduced_dim+7, y= np.cumsum(variances)[reduced_dim-1]-3, s="$d'=%d$\n%.2f %%" % (reduced_dim, np.cumsum(variances)[reduced_dim-1]), fontsize=16)
     
            plt.ylim([variances[0], np.sum(variances)])
            plt.tick_params(axis='x', labelsize=14)
            plt.tick_params(axis='y', labelsize=14)

            plt.savefig(data_dir+"pca_variance_ratios.png")
            plt.show()  
    
else:
    print("If full gene expression matrix is stored in the directory, directly apply PCA to the full matrix.")
    try:
        image = plt.imread(data_dir+"pca_variance_ratios.png")
        plt.imshow(image)
        plt.axis('off') 
        plt.show()
    except:
        print("data/Gene_Expression_Matrix.txt is needed!")
````

## Transport source to target using GPA
* Put ```--dataset Transport_genes -N_dim (d_red) --label (day1) (day2) --reduction (dim_red_method) --exp_no (exp_memo)``` and additional GPA training parameters that you want to temporarily change for this experiment. 
* Default GPA parameters and hyper-parameters are listed in ```configs/Transport_genes-GPA_NN.yaml```.
* Recommended tunable parameters depending on examples are ```--epochs``` for number of transportation steps, ```--lr_P``` for time step size for the ODE, ```--epochs_phi``` for number of gradient ascent updates for $\phi$, ```--lr_phi``` gradient ascent stepsize for $\phi$.
* For the loss function, 
    * the default setting is the Lipschitz regularized KL divergence in Donsker-Varadhan (DV) formulation; ```--f KL --formulation DV -L 1.0``` 
    * But if KL divergence is not suitable for the loss, we can try another Lipschitz regularized $f$-divergence in Legendre transform (LT) formulation; <br> ex) ```--f alpha -alpha 2.0 --formulation LT -L 1.0```
* Lipschitz constant $L$ is an algorithm level parameter. ```-L 1.0``` makes the algorithm stable in general. 
* `--` tag means required parameter, `-` tag means optional parameter defined in ```scripts/util/input_args.py```.
* The results will be saved in ```assets/Transport_genes/```. Results include 
    * movie for the evolving particles in the designated 2 axes ```plot_axes``` and ```plot_axes_join```
    * movie for the evolving vector field in the designated 2 axes ```plot_axes``` and ```plot_axes_join```
    * trajectories of particles in time in pickle format.

Steps for transportation
1. transport between consecutive time points
2. transport between not adjacent time points for prediction
3. transport using GPA with designed potential 
    - potential consits of nonlinear part (-> nonlinear residual dynamids) + quadratic part (-> linear dynamics)
    $$\phi(x, t) = \phi_{ML}(x, t) + \frac{1}{2} x^T A_t^TA_t x + b_t x$$
    - caution: Lipschitz regularized GPA would prefer quadratic terms and therefore some mollification/proximals is needed 
    - Assuming $A_t$ sparse, quadratic potential would not be critical

    - $A_t$ is given as prior knowledge which is constant in time, then we can update $A$ and also find how $A_t$ evolves in time 
    - needs validation
    
(4. traverse multiple time points -> beyond GPA)

````python
with open(result_filepath, "rb") as fr:
    param, result = pk.load(fr)
    X1_red_trpts = result['trajectories']
````

````python
def calculate_time_steps(dt, iter_nos, physical_time=True):
# calculate real_time / iteration count
    if physical_time == True:
        iter_nos_new = []
        if type(dt) == list: # decaying or varying dt
            for iter_no in iter_nos:
                if iter_no == 0:
                    iter_nos_new.append(0)
                else:
                    iter_nos_new.append(sum(dt[:iter_no]))
        else: # constant dt
            for iter_no in iter_nos:
                iter_nos_new.append(dt*iter_no)
        iter_nos = iter_nos_new
    return iter_nos
````

````python
# epochs(=iteration counts) -> physical time
ts = range(1, epochs+1)
print("Number of epochs: ", epochs, ", time step size: ", param['lr_P'])
pts = calculate_time_steps(param['lr_P'], ts)

X1_red_trpt = X1_red_trpts[-1]
````

### Performance metric
1. $D_f^{\Gamma_L}$ divergence : used for GPA <br>Divergence of reconstructed data and original data in the original space is bounded by the bounded by the same divergence in the latent space. (In PCA, the decoder is isometry. Therefore a metric calculated in the latent space is the same as the metric calculated in the reconstructed space.) <br> Note that the divergence is not symmetric on the order of datasets.
2. $W_2$ metric / sinkhorn divergence: objective metric <br>It is a metric!

````python
# Load f_Lip divergence calculating function
from scripts.GPA_NN.loss_NN import f_Lip_divergence
from functools import partial

f_Lip_div = partial(f_Lip_divergence, params={'f':f, 'L':1.0, 'alpha':alpha})
````

````python
# Load W2 calculating function
try:
    from geomloss import SamplesLoss
except:
    !pip3 install geomloss
    from geomloss import SamplesLoss
import torch
def sinkhorn(X, Y):
    X = torch.from_numpy(X).type(torch.float32)
    Y = torch.from_numpy(Y).type(torch.float32)

    return SamplesLoss(loss='sinkhorn', p=2)(X, Y).numpy()
````

````python
# PCA basis is orthonormal and so projection map is distance preserving
# Therefore distances between reconstructed data can be also calculated in the latent space
X1_trpt = pca.inverse_transform(X1_red_trpt)
#print(X1_trpt[:10, :10])
X1_hat = pca.inverse_transform(pca.transform(mats[day1]))
X2_hat = pca.inverse_transform(pca.transform(mats[day2]))
````
