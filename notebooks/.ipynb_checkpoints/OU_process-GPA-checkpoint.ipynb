# OU process (gradient flow for KL divergence with Gaussian stationary distribution) and its approximation through GPA

We compare particle trajectories of OU process from known drift term and diffusion term, and particle trajectories of GPA from  𝑁
  samples from the target.
  We claim that there exists a finite Lipschitz constant bound $L$ for the regularized KL-divergence where the ODE dynamics from Lipschitz-regularized GPA can recover the OU process.

```python
import numpy as np
import matplotlib.pyplot as plt
import pickle

np.random.seed(0)
N_dim = 2
```

```python
N_samples_P = 200
Y_means, Y_stds = [0.0, 4.0], [0.5, 1.0]
# p(y)= 1/3 * exp(-((y-Y_means[0])/Y_stds[0])**2/2)/(Y_stds[0]*sqrt(2*pi))+ \
#       2/3 * exp(-((y-Y_means[1])/Y_stds[1])**2/2)/(Y_stds[1]*sqrt(2*pi))
idx_0 = np.random.random((N_samples_P,1))>2/3
Y_0 = idx_0 * np.random.normal(loc = Y_means[0], scale = Y_stds[0], size=(N_samples_P, N_dim)) + \
(1-idx_0) * np.random.normal(loc = Y_means[1], scale = Y_stds[1], size=(N_samples_P, N_dim))
```

```python
N_samples_Q = 200
X_mean, X_std = 4.0, 1.0
drift = lambda x: -(x-X_mean)/X_std 
X_ = np.random.normal(loc = X_mean, scale = X_std, size=(N_samples_Q, N_dim))
```

## Dataset

* $M$ samples from a **source** distribution $P_0 = \frac{1}{3}N((0,0), 0.5^2I)+\frac{2}{3}N((4,4), I)$
* $N$ samples from a **target** distribution $Q = N((4,4), I)$ and its pdf

```python
plt.scatter(Y_0[:,0], Y_0[:,1], label='initial')
plt.scatter(X_[:,0], X_[:,1], label='target')
plt.legend(fontsize=12)
plt.show()
```

## OU process
Assume that the initial distribution is transported toward the target distribution through the gradient flow of the KL divergence or the Fokker-Planck equation.
$$\partial_t P_t - \nabla \cdot \left(P_t \nabla \left( \log \frac{dP_t}{dQ} \right)  \right) = 0, \,\,\, P_0 \text{ is given}$$
or
$$\partial_t P_t + \nabla \cdot \left(P_t \nabla \log Q  \right) = \Delta P_t,\,\,\, P_0 \text{ is given}.$$

Particle trajectories of OU process are obtained by $M$ initial samples and known drift term and diffusion term.
The corresponding OU process is
$$dX_t = \nabla \log Q(X_t) dt + \sqrt{2} dW_t$$
where $W_t$ denotes the Brownian motion
which is numerically calculated by Euler-Maruyama as
$$X_{t+1} = X_t + \nabla \log Q(X_t)dt + \sqrt{2}dW, \quad dW \sim N(0, dt * I).$$

We test on 200 initial samples from $P_0$ and $dt=0.02$, total number of iterations = 1000.
In our problem, $$\log Q(x) = -(x - (4,4))$$ where the speed linearly depends on the distance between the mean $(4,4)$.


```python
def OU_process(Y_0, drift, sigma, dt = 0.01, iterations=200):
    Ys = [Y_0]
    for i in range(iterations):
        Ys.append(Ys[-1] + drift(Ys[-1]) * dt + np.sqrt(2*dt)*sigma * np.random.normal(size=np.shape(Ys[-1])))
        
    return Ys
```

```python
dt = 0.02
sigma = 1.0
iterations = 1000
Ys = OU_process(Y_0, drift, sigma, dt, iterations=iterations)
steps = 5
Ys_steps = Ys[0:-1:steps]
exp_no = 1
```

```python
# save OU samples
filename = data_dir + "ou_result_%s.pickle" % str(exp_no)
with open(filename,"wb") as fw:
    pickle.dump([Ys_steps[-1]] , fw)
print("Results saved at:", filename)
animation_filename = result_dir + "ou_iterations_%s.gif" % str(exp_no)
```

```python
from matplotlib import animation

# First set up the figure, the axis, and the plot element we want to animate
fig, ax = plt.subplots()
ax.set_xlim([-1.5, 7])
ax.set_ylim([-1.5, 7])

scats = []

ttl = ax.text(0.5,1.05, "t = %.3f" % 0.0,\
              bbox={'facecolor':'w', 'alpha':0.5, 'pad':5},\
              transform=ax.transAxes, ha="center")
scat = ax.scatter(Y_0[:,0], Y_0[:,1], c='tab:blue')#, label='initial')
ax.scatter(X_[:,0], X_[:,1], c='tab:orange')#, label='target')

scats.append([scat, ttl])

for i, Y in enumerate(Ys_steps):
    ttl = ax.text(0.5,1.05, "t = %.3f" %((i+1)*dt*steps),\
              bbox={'facecolor':'w', 'alpha':0.5, 'pad':5},\
              transform=ax.transAxes, ha="center")
    scat = ax.scatter(Y[:,0], Y[:,1], c='tab:blue')#, label='transported')
    
    scats.append([scat, ttl])

ani = animation.ArtistAnimation(fig, scats, interval=50, blit=True, repeat_delay=200)
writergif = animation.PillowWriter(fps=3)
ani.save(animation_filename , writer=writergif)
```

## GPA 

In GPA, the particles are transported through the ODE 
$$\frac{dX_t}{dt} = -\nabla\phi_t(X_t) \approx \nabla\left( \log \frac{dP_t}{dQ}  \right)(X_t)$$
where $\phi_t$ is estimated by $M$ samples $Y_t$ and $N$ target samples $X$.

We test on 200 initial samples from $P_0$ and $dt=0.02$, total number of iterations = 1000.

The speed bound is set to $L=10$ to catch up the OU dynamics.


```python
L =  1.0
result_filename = result_dir + 'KL-Lipschitz_%.4f-0200_0200-00-%s.pickle' % (L, str(exp_no))


if not os.path.exists(result_filename):
    os.chdir("scripts/GPA")

    cmd = f"python3 train.py --dataset Toy1 --config GPA-Toy1 -L {L} --exp_no {exp_no} --epochs 1000 --adjust_lr_P_epochs N"
    !{cmd}

    os.chdir(main_dir)

with open(result_filename, "rb") as fr:
    param_1, result_1 = pickle.load(fr)

filename = result_dir + 'KL-Lipschitz_%.4f-0200_0200-00-%s-0_1-movie.gif' % (L, str(exp_no))
display(Image(filename = filename))

```

```python
L = 10.0
result_filename = result_dir + 'KL-Lipschitz_%.4f-0200_0200-00-%s.pickle' % (L, str(exp_no))


if not os.path.exists(result_filename):
    os.chdir("scripts/GPA")
    
    cmd = f"python3 train.py --dataset Toy1 --config GPA-Toy1 -L {L} --exp_no {exp_no} --f_Lip_threshold 1e-6 --epochs 1000"
    !{cmd}
    
    os.chdir(main_dir)

with open(result_filename, "rb") as fr:
    param_10, result_10 = pickle.load(fr)
        
filename = result_dir + 'KL-Lipschitz_%.4f-0200_0200-00-%s-0_1-movie.gif' % (L, str(exp_no))

display(Image(filename = filename))

```

## Compare the trajectories from the OU process and GPA
Calculate Wasserstein distance (an objective metric) between $P_t$ and $Q$ along the OU process trajectory and the GPA trajectory


```python
# Load W2 calculating function
try:
    from geomloss import SamplesLoss
except:
    !pip3 install geomloss
    from geomloss import SamplesLoss
import torch
def W2(X, Y):
    X = torch.from_numpy(X).type(torch.float32)
    Y = torch.from_numpy(Y).type(torch.float32)

    return SamplesLoss(loss='sinkhorn', p=2)(X, Y).numpy()
```

```python
W2_OU_target = [W2(Y_, X_) for Y_ in Ys_steps]
print("done with OU trajectories")

W2_GPA_1_target = [W2(Y_, X_) for Y_ in result_1['trajectories']]
print("done with KL-Lip=1 trajectories")

W2_GPA_10_target = [W2(Y_, X_) for Y_ in result_10['trajectories']]
print("done with KL-Lip=10 trajectories")
```

```python
len(W2_GPA_10_target)
```

```python
t = np.arange(0, dt*iterations, dt*steps)
plt.semilogy(t, W2_OU_target, label = 'OU process')
plt.semilogy(t, W2_GPA_1_target, 'r--', label = 'KL-Lip=1 GPA', linewidth=2)
plt.semilogy(t, W2_GPA_10_target, 'k--', label = 'KL-Lip=10 GPA',  linewidth=2)
plt.legend(fontsize=12)
plt.title(r'$W_2(P_t, Q)$,  $Q$=stationary distribution', fontsize=12)
plt.xlabel(r'$t$')
plt.savefig(result_dir+"convergence.png")
plt.show()
```
