## Stem Cell Differentiation dataset

### Data property

* 5 snapshots at different timepoints (day 0, 1, 2, 3, 4)
* samplesize: differ by days
* dimension: 101

### Data structure
Datasets are located in ```data/stem_Cell_Differentiation``` directory. 


* ```Data_0_gene_expression_matrix.txt``` : Raw data matrix in $\mathbb{R}^{M \times N}$ is composed of $M$ rows for individual samples and $N$ columns of different genes.
* ```Data_0_Cells_Days_matrix.txt``` : The entire dataset is divided into 6 classes which correspond to different days.


````python
# Preprocessing function
def load_stem_cell_diff_dataset():
    import pandas as pd

    # gene_expression dataset => full_matrix
    GE_matrix_file = data_dir + 'Stem_Cell_Differentiation/Data_0_gene_expression_matrix.txt'
    if os.path.exists(GE_matrix_file):
        df = pd.read_table(GE_matrix_file, sep="\t")
        print(df)
        full_matrix = np.array(df)[:,1:]
    else:
        print("If full gene expression matrix is stored in the directory, load and view it.")

    # time information => time_label
    time_info_file = data_dir + 'Stem_Cell_Differentiation/Data_0_Cells_Days_matrix.txt'
    df_cls = pd.read_table(time_info_file, sep="\t")
    time_label = np.array(df_cls['day'])
    print("classes:", set(time_label))
        
    return full_matrix, time_label
````

````python
# Data preprocess
from scripts.util.data_preparations import *
from scripts.util.visualizations import visualize_data

example_name = 'stem_cell_differentiation' # 
d_reds = [2, 4, 8, 16,32, 64,101] # [] or list of reduced dimensions
    
full_matrix, time_label = load_stem_cell_diff_dataset()
visualize_data(full_matrix, time_label, colors=contrast_colors)
try:
    pca, projected_matrix = reduce_dimension(full_matrix, d_reds, example_name)
    data = {'full_matrix': full_matrix, 'time_label': time_label, 'pca': pca, 'projected_matrix': projected_matrix}
except:
    data = {'full_matrix': full_matrix, 'time_label': time_label}
save_preprocessed_data(data, example_name)
````

## PROFET

````python
# PROFET step 1: GPA
d_red = 2
times = [0, 2, 4]
dimension_reduction = 'Y'
exp_no = "test"

# tunable paramters
adjust_lr_P_epochs = 'Y'    # 'Y' for adjusting lr_P w.r.t. the size of data (source+target) support and epochs w.r.t. dimensionality
                            # 'N' for using specified lr_P and epochs below
lr_P = 0.0025               
epochs = 5000
specify_f_Lip_threshold = 'Y'    # 'Y' for stopping GPA when f_Lip divergence gets lower than the specified threshold below
f_Lip_threshold = 1e-4           # 'N' for using default threshold in configs/GPA.Yaml if exists or setting not to stop early
````

````python
os.chdir("scripts/GPA")

for i in range(len(times)-1):
    cmd = f"python3 train.py --dataset {example_name} --label {times[i]} {times[i+1]} --N_dim {d_red} --exp_no {exp_no} --dimension_reduction {dimension_reduction}"
    
    if adjust_lr_P_epochs in ['N', 'no']:
        cmd += f" --adjust_lr_P_epochs {adjust_lr_P_epochs} -lr_P {lr_P} --epochs {epochs}"
    if specify_f_Lip_threshold in ['Y', 'yes']:
        cmd += f" --f_Lip_threshold {f_Lip_threshold}"
    
    !{cmd}

os.chdir(main_dir)
````

````python
# PROFET step 2: Force-Matching
intermediate_times = [1, 3]

exp_memo = f"dim{d_red}_test"
if dimension_reduction in ['Y', 'yes']:
    dimension_note = f'dim{d_red}-'
else:
    dimension_note = ''
gpa_filenames = [f'KL-Lipschitz_1.0000-{times[0]}_{times[1]}times-{dimension_note}{cnt[times[1]]:04d}_{cnt[times[0]]:04d}-00-{exp_no}.pickle',
                f'KL-Lipschitz_1.0000-{times[1]}_{times[2]}times-{dimension_note}{cnt[times[2]]:04d}_{cnt[times[1]]:04d}-00-{exp_no}.pickle',]

gpa_filenames_join = " ".join(gpa_filenames)
times_join = " ".join([str(t) for t in times])
````

````python
os.chdir("scripts/ForceMatching")
cmd = f"python3 train.py --dataset {example_name} --ts {times_join} --exp_memo {exp_memo} --files {gpa_filenames_join}" 

!{cmd}

os.chdir(main_dir)
````

````python
from scripts.ForceMatching.evaluations import load_velocityfield, time_integration

# load velocityfield
velocity_net, p = load_velocityfield(f"assets/{example_name}/{exp_memo}/")

# integrate ODE
dt = p['numerical_ts'][-1]/200
numerical_dt = dt
physical_dt = dt * p['ts'][-1] / p['numerical_ts'][-1]
input_data = full_matrix[time_label==0,:] if dimension_reduction in ['N', 'no'] else projected_matrix[time_label==0, :d_red]
X1_trpts = time_integration(input_data, T = p['numerical_ts'][-1], dt = dt, velocity_net=velocity_net)
````

````python
from scripts.util.visualizations import animate_trajectories, plot_w2errors

img_src1 = f"assets/{example_name}/{exp_memo}-movie-particles.gif"
img_src2 = f"assets/{example_name}/{exp_memo}-movie-velocities.gif"
img_src3 = f"assets/{example_name}/{exp_memo}-w2distances.png"

animate_trajectories(X1_trpts, times, intermediate_times, \
                     time_label, projected_matrix, \
                     img_src1, numerical_dt, physical_dt=physical_dt, \
                     colors=contrast_colors, dimension_reduction=True, plot_vectorfield=False)
animate_trajectories(X1_trpts, times, intermediate_times, \
                     time_label, projected_matrix, \
                     img_src2, numerical_dt, physical_dt=physical_dt, \
                     colors=contrast_colors, dimension_reduction=True, plot_vectorfield=True)
reference_matrix = full_matrix if dimension_reduction in ['N', 'no'] else projected_matrix[:, :d_red]
all_times = sorted(times + intermediate_times)
plot_w2errors(X1_trpts, time_label, all_times, img_src3, reference_matrix, physical_dt, colors=contrast_colors)
````
