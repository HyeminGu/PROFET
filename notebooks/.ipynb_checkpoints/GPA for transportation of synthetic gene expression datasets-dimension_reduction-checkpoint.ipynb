# GPA for transportation of synthetic gene expression datasets

This notebook aims to apply GPA to recover the trajectory of a synthetic gene expression dataset which is produced by known (deterministic) dynamics.

## Dataset

* 5 snapshots at different timepoints (day 0 to day 4) 
* samplesize : 239 
* dimenison : 26
* The trajectory will be visualized in 2D principal component axes

## Functionality

* Given source and target days, GPA transports the source dataset toward the target dataset from a deterministic particle dynamics for the gradient flow of (Lipschitz regularized or limited transportation speed) KL divergergence.
$$D_{KL}^L(P\|Q) = \sup_{\| \nabla \phi \| \leq L} \left \{\mathbb{E}_P[\phi] - \log \mathbb{E}_Q [\exp(\phi)] \right \}$$
$$\partial_t P + \nabla \cdot \left(P v\right) = \partial_t P - \nabla \cdot \left(P \nabla \phi \right) = 0$$
$$\dot{X} = - \nabla \phi_t(X)$$

* Snapshots at all the intermediate points will be highlighted by default, but it can be specified.
* $W_2$ distance will be calculated between the particle trajectory and the snapshots.



## Load and preprocess data

Datasets are located in ```data/traj_dataset/log_traj_#.log``` directory. The trajectory will be visualized in 2D principal component axes.


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pickle as pk

from IPython.display import Image, display

import os
import sys

main_dir = os.getcwd()
if "notebooks" in main_dir:
    main_dir = main_dir[:-10]
    os.chdir(main_dir)
elif main_dir == "/content":
    main_dir = main_dir + "/GPA-source_code"
print(main_dir)
    
sys.path.append(main_dir)
data_dir = main_dir + '/data/'
````

````python
cls = [0, 1, 2, 3, 4]
N_samples_cls = {}
mats = {}
for c in cls:
    if os.path.exists(data_dir + 'traj_dataset/log_traj_%d.log' % c):
        df = pd.read_table(data_dir + 'traj_dataset/log_traj_%d.log' % c, sep="\t", header=None)
        #print(df)
        mats[c] = np.array(df)[:,:-1]
    else:
        print("Missing data at day %d." % c)
    N_samples_cls[c] = mats[c].shape[0]
    print("Day ", c, ": sample size: ", N_samples_cls[c])

 
````

````python
from sklearn import decomposition
````

````python
pca = decomposition.PCA(n_components=2, random_state=0)
mat = np.concatenate([mats[c] for c in cls])
pca.fit(mat)
vis_all_days = pca.transform(mat)
````

````python
np.random.seed(127)
colors = {}

for day in cls:
    vis_day = pca.transform(mats[day])
    colors[day] = np.random.random(3)
    
    fig = plt.figure(figsize=(4,3))
    plt.scatter(vis_all_days[:,0], vis_all_days[:,1], color='lightgray', alpha=0.3, s=0.7)
    plt.scatter(vis_day[:,0], vis_day[:,1], alpha=1.0, color=colors[day], label = f'day{day}', s=0.7)
    plt.legend()
    plt.show()
````

## Transport source to target using GPA
* Put ```--dataset Transport_genes -N_dim (d_red) --label (day1) (day2) --reduction (dim_red_method) --exp_no (exp_memo)``` and additional GPA training parameters that you want to temporarily change for this experiment. 
* Default GPA parameters and hyper-parameters are listed in ```configs/Transport_genes-GPA_NN.yaml```.
* Recommended tunable parameters depending on examples are ```--epochs``` for number of transportation steps, ```--lr_P``` for time step size for the ODE, ```--epochs_phi``` for number of gradient ascent updates for $\phi$, ```--lr_phi``` gradient ascent stepsize for $\phi$.
* For the loss function, 
    * the default setting is the Lipschitz regularized KL divergence in Donsker-Varadhan (DV) formulation; ```--f KL --formulation DV -L 1.0``` 
    * But if KL divergence is not suitable for the loss, we can try another Lipschitz regularized $f$-divergence in Legendre transform (LT) formulation; <br> ex) ```--f alpha -alpha 2.0 --formulation LT -L 1.0```
* Lipschitz constant $L$ is an algorithm level parameter. ```-L 1.0``` makes the algorithm stable in general. 
* `--` tag means required parameter, `-` tag means optional parameter defined in ```scripts/util/input_args.py```.
* The results will be saved in ```assets/Transport_genes/```. Results include 
    * movie for the evolving particles in the designated 2 axes ```plot_axes``` and ```plot_axes_join```
    * movie for the evolving vector field in the designated 2 axes ```plot_axes``` and ```plot_axes_join```
    * trajectories of particles in time in pickle format.
    
The video of resulting GPA trajectory in the latent space is stored in ```assets/Transport_genes/``` directory as gif format file. 

````python
# Wasserstein-2 distance
import torch # for W2 calculation purpose
try:
    from geomloss import SamplesLoss
except:
    !pip3 install geomloss
    from geomloss import SamplesLoss

def W2(X, Y):
    X = torch.from_numpy(X).type(torch.float32)
    Y = torch.from_numpy(Y).type(torch.float32)

    return SamplesLoss(loss='sinkhorn', p=2)(X, Y).numpy()


# conversion from gpa iterations to algorithm time 
def calculate_time_steps(dt, iter_nos, physical_time=True):
    if physical_time == True:
        iter_nos_new = []
        if type(dt) == list: # decaying or varying dt
            for iter_no in iter_nos:
                if iter_no == 0:
                    iter_nos_new.append(0)
                else:
                    iter_nos_new.append(sum(dt[:iter_no]))
        else: # constant dt
            for iter_no in iter_nos:
                iter_nos_new.append(dt*iter_no)
        iter_nos = iter_nos_new
    return iter_nos


import matplotlib.animation as animation
````

````python
# function to apply GPA
def run(source_t, target_t, intermediate_t = [], 
        total_iter = 10000, termination_error = 0.01, termination_error_metric = 'KE', 
        exp_memo = '2'):
    intermediate_t = np.array(intermediate_t)
    if len(intermediate_t) == 0:
        intermediate_t = range(source_t+1, target_t)
        
    # data parameters
    day1, day2 = source_t, target_t

    # GPA parameters
    f, alpha, L = 'KL', None, 1.0
    #f, alpha, L = 'alpha', 2.0, 1.0
    epochs = total_iter # number of updates
    lr_P = 0.1 # time stepsize
    plot_axes = '0 1' # 2 axes for plotting in 2D
    plot_axes_join = '0_1'

    # --------
    N_source = N_samples_cls[day1]
    N_target = N_samples_cls[day2]

    result_dir = '%s/assets/Test_transport_genes/' % main_dir
    if f == 'KL':
        f_Lip = 'KL-Lipschitz_%.4f' % L
    else:
        f_Lip = 'alpha=%05.2f-Lipschitz_%.4f' % (alpha, L)
    result_name = f_Lip + '-%d_%dtimes_%04d_%04d_00_%s' % \
    (day1, day2, N_target, N_source, exp_memo)
    print(result_name)
    result_filepath = result_dir + result_name + '.pickle'
    
    
    # run GPA
    if os.path.exists(result_filepath): # load existing result
        with open(result_filepath, "rb") as fr:
                param, result = pk.load(fr)
                X1_trpts = result['trajectories']
    else:  # run GPA
        if f == 'KL':
            !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $day1 $day2 \
            --exp_no $exp_memo --f $f --formulation DV --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P\
            -termination_error $termination_error -termination_error_metric $termination_error_metric
        else:
            !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $day1 $day2 \
            --exp_no $exp_memo --f $f -alpha $alpha --formulation LT --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P\
            -termination_error $termination_error -termination_error_metric $termination_error_metric
    
    
    #img_src = result_dir + result_name + '-%s-movie.gif' % plot_axes_join
    #display(Image(filename = img_src))

    
    # read the result
    with open(result_filepath, "rb") as fr:
        param, result = pk.load(fr)
        X1_trpts = result['trajectories']
        
    # epochs(=iteration counts) -> physical time
    ts = np.arange(1, len(result['trajectories'])*param['save_iter']+1, param['save_iter'])
    print("Number of epochs: ", len(result['trajectories'])*param['save_iter'], ", time step size: ", param['lr_P'])
    pts = param['lr_P'] * ts

    X1_trpt = X1_trpts[-1]
    
    
    
    # Visualization in the original space (UMAP)
    img_src = result_dir + result_name + '-movie-original.gif' 
    if os.path.exists(img_src): # try loading saved movie
        display(Image(filename = img_src))
    else: # draw, save and load
        fig, ax = plt.subplots()
        ims = []
        X1_vis = pca.transform(mats[day1])
        X2_vis = pca.transform(mats[day2])
        #X1_intermediate_vis = pca.transform(mats[day1+1])
        im = ax.scatter(X1_vis[:,0], X1_vis[:,1], color=colors[day1], 
                        alpha=1.0, s=0.7, zorder=10, label=f'day {day1}') # source in red
        ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[day2], 
                   alpha=1.0, s=0.7, zorder=10, label=f'day {day2}') # target in blue
        for t in intermediate_t:
            X1_intermediate_vis = pca.transform(mats[t])
            ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color=colors[t], 
                        alpha=1.0, s=0.7, zorder=10, label=f'day {t}')
        ax.scatter(vis_all_days[:,0], vis_all_days[:,1], color='lightgray', 
                   alpha=0.3, s=0.7, zorder=1) # backgrounds in gray
        #leg = ax.legend(loc='upper right')
        ttl = ax.text(0.5,1.05, "t = %.3f" % 0.0, \
                      bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                      transform=ax.transAxes, ha="center")
        ims.append([im, ttl])#, leg])
        for i, X1_trpt in enumerate(X1_trpts):  # trajectories
            if np.isnan(X1_trpt).any():
                break
            X1_hat_vis = pca.transform(X1_trpt)
            im = ax.scatter(X1_hat_vis[:,0], X1_hat_vis[:,1], color=colors[day1], 
                            alpha=1.0, s=0.7, zorder=10, label=f'day {day1}') # transported source 
            ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[day2], 
                       alpha=1.0, s=0.7, zorder=10, label=f'day {day2}') # target
            ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color=colors[day1+1], 
                        alpha=1.0, s=0.7, zorder=10, label=f'day {day1+1}')
            ax.scatter(vis_all_days[:,0], vis_all_days[:,1], color='lightgray', 
                       alpha=0.3, s=0.7, zorder=1) # backgrounds in gray
            #leg = ax.legend(loc='upper right')
            ttl = ax.text(0.5,1.05, "t = %.3f" % pts[i], \
                          bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                          transform=ax.transAxes, ha="center")
            ims.append([im, ttl])#, leg])
        ani = animation.ArtistAnimation(fig, ims, interval=50, blit=True, repeat_delay=200)
        writergif = animation.PillowWriter(fps=3)
        ani.save(img_src, writer=writergif)
        plt.clf()
        display(Image(filename = img_src))
        


    # compare W2 error with intermediate timepoints
    img_src2 = result_dir + result_name + '-w2distances.png' 
    if os.path.exists(img_src2): # try loading saved movie
        display(Image(filename = img_src2))
    else:
        W2_s = []
        all_ts = np.concatenate(([source_t], intermediate_t, [target_t]))
        for i, X1_trpt in enumerate(X1_trpts): # trajectories
            W2_s.append([W2(X1_trpt, mats[day]).item() for day in all_ts])
            #print(f"{ts[i]}: {W2_s[j][i]}")

        plt.plot(param['lr_P']*np.arange(0, len(W2_s)*param['save_iter'], param['save_iter']), W2_s)
        plt.legend(["Day" + str(x) for x in all_ts])
        plt.xlabel("GPA time")

        argmins = np.argmin(np.array(W2_s), axis=0) * param['save_iter'] * param['lr_P']
        #for m in argmins:
        #    plt.axvline(x=m, c='red')
        print("minimum values are obtained at x=", argmins)

        plt.title(r'$W_2(P_n, Q_{day})$')  
        plt.savefig(img_src2)
        plt.show()

        
````

## Mixture GPA

Here we define a time dependent target which is obtained as a mixture of transient targets and the stationary target.

$$Q_{t} = \sum_{i=1}^n c_{i}(t) \delta_{t\leq t_i} Q_i$$
for $i=1, \cdots, n$ where $Q_1, \cdots, Q_n$ are given snapshots and $c_i(t)$ is piecewise constant in time i.e. constant for $t_i \leq t \leq t_{i+1}$.

````python
# function to apply GPA
def run_mixture(source_t, target_t, intermediate_t = [], 
                iter_transient_target = [], total_iter = 10000,
                termination_error = 0.01, termination_error_metric = 'KE', 
                weight_decay_mixture_target=1.0,
                exp_memo = '2'):
    intermediate_t = np.array(intermediate_t)
    if len(intermediate_t) == 0:
        intermediate_t = range(source_t+1, target_t)
        
    # data parameters
    if type(target_t) == type([]):
        days = [source_t] + target_t
    else:
        days = [source_t, target_t]
    days_join =  " ".join(str(l) for l in days)

    # GPA parameters
    f, alpha, L = 'KL', None, 1.0
    #f, alpha, L = 'alpha', 2.0, 1.0
    epochs = total_iter # number of updates
    lr_P = 0.1 # time stepsize
    plot_axes = '0 1' # 2 axes for plotting in 2D
    plot_axes_join = '0_1'

    # --------
    N_source = N_samples_cls[days[0]]
    N_target = 0
    for day in days[1:]:
        N_target += N_samples_cls[day]

    result_dir = '%s/assets/Test_transport_genes/' % main_dir
    if f == 'KL':
        f_Lip = 'KL-Lipschitz_%.4f' % L
    else:
        f_Lip = 'alpha=%05.2f-Lipschitz_%.4f' % (alpha, L)
    labels = "_".join(str(l) for l in days)
    
    result_name = f_Lip + '-%stimes_%04d_%04d_00_%s' % \
    (labels, N_target, N_source, exp_memo)
    print(result_name)
    result_filepath = result_dir + result_name + '.pickle'
    
    
    # run GPA
    if os.path.exists(result_filepath): # load existing result
        with open(result_filepath, "rb") as fr:
                param, result = pk.load(fr)
                X1_trpts = result['trajectories']
    else:  # run GPA
        if len(iter_transient_target) == 0: # automatically set iterations to hit transient target
            if f == 'KL':
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f --formulation DV --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P \
                -termination_error $termination_error -termination_error_metric $termination_error_metric \
                -weight_decay_mixture_target $weight_decay_mixture_target
            else:
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f -alpha $alpha --formulation LT --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P\
                -termination_error $termination_error -termination_error_metric $termination_error_metric \
                -weight_decay_mixture_target $weight_decay_mixture_target
                
        else:
            iter_transient_target_join = " ".join(str(l) for l in iter_transient_target)
            if f == 'KL':
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f \
                --formulation DV --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P -iter_transient_target $iter_transient_target \
                -termination_error $termination_error -termination_error_metric $termination_error_metric \
                -weight_decay_mixture_target $weight_decay_mixture_target
            else:
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f -alpha $alpha --formulation LT --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P\
                -termination_error $termination_error -termination_error_metric $termination_error_metric \
                -iter_transient_target $iter_transient_target \
                -weight_decay_mixture_target $weight_decay_mixture_target
    
    img_src = result_dir + result_name + '-%s-movie.gif' % plot_axes_join
    display(Image(filename = img_src))

    
    # read the result
    with open(result_filepath, "rb") as fr:
        param, result = pk.load(fr)
        X1_trpts = result['trajectories']
        
    # epochs(=iteration counts) -> physical time
    ts = range(1, epochs+1)
    print("Number of epochs: ", epochs, ", time step size: ", param['lr_P'])
    pts = calculate_time_steps(param['lr_P'], ts)

    X1_trpt = X1_trpts[-1]
    
    
    
    # Visualization in the original space (UMAP)
    img_src = result_dir + result_name + '-movie-original.gif' 
    if os.path.exists(img_src): # try loading saved movie
        display(Image(filename = img_src))
    else: # draw, save and load
        fig, ax = plt.subplots()
        ims = []
        X1_vis = pca.transform(mats[days[0]])
        im = ax.scatter(X1_vis[:,0], X1_vis[:,1], color=colors[days[0]], 
                        alpha=1.0, s=0.7, zorder=10, label=f'day {days[0]}') # source 
        
        for t in days[1:]:
            X2_vis = pca.transform(mats[t])
            ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[t], 
                   alpha=1.0, s=0.7, zorder=10, label=f'day {t}') # target 
            
        for t in intermediate_t:
            X1_intermediate_vis = pca.transform(mats[t])
            ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color='lightgray', 
                        alpha=0.3, s=0.7, zorder=1, label=f'day {t}')
        #ax.scatter(vis_all_days[:,0], vis_all_days[:,1], color='lightgray', 
        #           alpha=0.3, s=0.7, zorder=1) # backgrounds in gray
        #leg = ax.legend(loc='upper right')
        ttl = ax.text(0.5,1.05, "t = %.3f" % 0.0, \
                      bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                      transform=ax.transAxes, ha="center")
        ims.append([im, ttl])#, leg])
        for i, X1_trpt in enumerate(X1_trpts):  # trajectories
            if np.isnan(X1_trpt).any():
                break
            X1_hat_vis = pca.transform(X1_trpt)
            im = ax.scatter(X1_hat_vis[:,0], X1_hat_vis[:,1], color=colors[days[0]], 
                            alpha=1.0, s=0.7, zorder=10, label=f'day {days[0]}') # transported source 
            for t in days[1:]:
                X2_vis = pca.transform(mats[t])
                ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[t], 
                   alpha=1.0, s=0.7, zorder=10, label=f'day {t}') # target 
            
            for t in intermediate_t:
                X1_intermediate_vis = pca.transform(mats[t])
                ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color='lightgray', 
                        alpha=0.3, s=0.7, zorder=1, label=f'day {t}')
            
            #leg = ax.legend(loc='upper right')
            ttl = ax.text(0.5,1.05, "t = %.3f" % pts[i], \
                          bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                          transform=ax.transAxes, ha="center")
            ims.append([im, ttl])#, leg])
        ani = animation.ArtistAnimation(fig, ims, interval=50, blit=True, repeat_delay=200)
        writergif = animation.PillowWriter(fps=3)
        ani.save(img_src, writer=writergif)
        plt.clf()
        display(Image(filename = img_src))
        


    # compare W2 error with intermediate timepoints
    img_src2 = result_dir + result_name + '-w2distances.png' 
    if os.path.exists(img_src2): # try loading saved movie
        display(Image(filename = img_src2))
    else:
        W2_s = []
        if type(target_t) == type([]):
            all_ts = np.concatenate(([source_t], intermediate_t, target_t))
        else:
            all_ts = np.concatenate(([source_t], intermediate_t, [target_t]))
        all_ts = np.sort(all_ts)

        for i, X1_trpt in enumerate(X1_trpts): # trajectories
            W2_s.append([W2(X1_trpt, mats[day]).item() for day in all_ts])
            #print(f"{ts[i]}: {W2_s[j][i]}")

        plt.plot(range(0, epochs, param['save_iter']), W2_s)
        plt.legend(["Day" + str(x) for x in all_ts])

        argmins = np.argmin(np.array(W2_s), axis=0) * param['save_iter']
        print("minimum values are obtained at x=", argmins)

        plt.title(r'$W_2(P_n, Q_{day})$')  
        plt.savefig(img_src2)
        plt.show()
    
        
````

* ```source_t``` ; integer scalar: source day
* ```target_t``` ; integer list: list of target days that are informed to GPA
* ```intermediate_t```; integer list: list of intermediate days that are not informed to GPA but is wanted to be verified
* ```iter_transient_target```; integer list: list of iteration counts to reach intermediate targets
* ```total_iter```; integer scalar: iteration count to reach the target
* ```weight_decay_mixture_target```; float scalar: weight decay rate in time >=0 in the exponent, equal rate for all time for 0, and faster decaying rate as time goes for larger value
* ```exp_memo```; string: naming on the experiment run

````python
source_t, target_t = 0, [2, 4]
run_mixture(source_t, target_t, intermediate_t = [1,3],
            iter_transient_target = [5000], total_iter = 10000,
            termination_error = 0.01, termination_error_metric = 'f_Lip', # termination_error_metric from ['f_Lip', 'KE']
            weight_decay_mixture_target=1.0,
            exp_memo = '3')
````

## Regularized GPA

Here we assume that the ODE for the particle dynamics is splitted into gradient and non-gradient part where the non-gradient part is linear and pre-calculated from data.

$$\dot{x} = - \nabla \phi(x) + C(Ax +b)$$

$$x_{t+1} = x_t - \lambda_\phi \nabla \phi(x_t) + \lambda_{A,b}(Ax_t +b)$$

Currently, piecewise-constant nongradient drift (piecewise constant in time $t_i \leq t \leq t_{i+1}$) is available with ```generative_model```='GPA_NN_w_simple_nongradient_drift'.
In order to make a dictionary of nongradient drift, go to ```scripts/``` folder in your terminal and run ```calc_simple_nongradient_drift.py``` with
```bash
./calc_simple_nongradient_drift.py Test_transport_genes
```
where the resulting vectorfields with (```N_dim```+2) columns for \[source day, target day, average displacements from source to target\] will be stored in ```data/Test_transport_genes_simple_avg_directions_dim26.csv```.
Then run the function below ```run_regularized()``` with parameters described in the later cell.

````python
# function to apply GPA
def run_regularized(source_t, target_t, intermediate_t = [], 
                    iter_transient_target = [], total_iter = 10000,
                    termination_error = 0.01, termination_error_metric = 'KE', 
                    generative_model='GPA_NN_w_simple_nongradient_drift',
                    ratio_nongrad_2_grad = 1.0,
                    exp_memo = '2'):
    intermediate_t = np.array(intermediate_t)
    if len(intermediate_t) == 0:
        intermediate_t = range(source_t+1, target_t)
        
    # data parameters
    if type(target_t) == type([]):
        days = [source_t] + target_t
    else:
        days = [source_t, target_t]
    days_join =  " ".join(str(l) for l in days)

    # GPA parameters
    f, alpha, L = 'KL', None, 1.0
    #f, alpha, L = 'alpha', 2.0, 1.0
    epochs = total_iter # number of updates
    lr_P = 0.1 # time stepsize
    plot_axes = '0 1' # 2 axes for plotting in 2D
    plot_axes_join = '0_1'

    # --------
    N_source = N_samples_cls[days[0]]
    N_target = 0
    N_target += N_samples_cls[days[-1]]

    result_dir = '%s/assets/Test_transport_genes/' % main_dir
    if f == 'KL':
        f_Lip = 'KL-Lipschitz_%.4f' % L
    else:
        f_Lip = 'alpha=%05.2f-Lipschitz_%.4f' % (alpha, L)
    labels = "_".join(str(l) for l in days)
    
    result_name = 'Regularized_' + f_Lip + '-%stimes_%04d_%04d_00_%s' % \
    (labels, N_target, N_source, exp_memo)
    print(result_name)
    result_filepath = result_dir + result_name + '.pickle'
    
    
    # run GPA
    if os.path.exists(result_filepath): # load existing result 
        with open(result_filepath, "rb") as fr:
                param, result = pk.load(fr)
                X1_trpts = result['trajectories']
    else:  # run GPA
        if len(iter_transient_target) == 0: # automatically set iterations to hit transient target
            if f == 'KL':
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f \
                --formulation DV --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P \
                -termination_error $termination_error -termination_error_metric $termination_error_metric\
                --generative_model $generative_model -ratio_nongrad_2_grad $ratio_nongrad_2_grad
            else:
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f -alpha $alpha \
                --formulation LT --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P \
                -termination_error $termination_error -termination_error_metric $termination_error_metric\
                --generative_model $generative_model -ratio_nongrad_2_grad $ratio_nongrad_2_grad
                
        else:
            iter_transient_target_join = " ".join(str(l) for l in iter_transient_target)
            if f == 'KL':
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f \
                --formulation DV --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P \
                -termination_error $termination_error -termination_error_metric $termination_error_metric\
                --generative_model $generative_model -iter_transient_target $iter_transient_target_join \
                -ratio_nongrad_2_grad $ratio_nongrad_2_grad
            else:
                !python3 $main_dir/main.py --dataset Test_transport_genes -L $L --label $days_join \
                --exp_no $exp_memo --f $f -alpha $alpha \
                --formulation LT --epochs $epochs -plot_axes $plot_axes --lr_P $lr_P\
                -termination_error $termination_error -termination_error_metric $termination_error_metric\
                --generative_model $generative_model -iter_transient_target $iter_transient_target_join \
                -ratio_nongrad_2_grad $ratio_nongrad_2_grad
    
    img_src = result_dir + result_name + '-%s-movie.gif' % plot_axes_join
    display(Image(filename = img_src))

    
    # read the result
    with open(result_filepath, "rb") as fr:
        param, result = pk.load(fr)
        X1_trpts = result['trajectories']
        
    # epochs(=iteration counts) -> physical time
    ts = range(1, epochs+1)
    print("Number of epochs: ", epochs, ", time step size: ", param['lr_P'])
    pts = calculate_time_steps(param['lr_P'], ts)

    X1_trpt = X1_trpts[-1]
    
    
    
    # Visualization in the original space (UMAP)
    img_src = result_dir + result_name + '-movie-original.gif' 
    if os.path.exists(img_src): # try loading saved movie
        display(Image(filename = img_src))
    else: # draw, save and load
        fig, ax = plt.subplots()
        ims = []
        X1_vis = pca.transform(mats[days[0]])
        im = ax.scatter(X1_vis[:,0], X1_vis[:,1], color=colors[days[0]], 
                        alpha=1.0, s=0.7, zorder=10, label=f'day {days[0]}') # source 
        
        for t in days[1:]:
            X2_vis = pca.transform(mats[t])
            ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[t], 
                   alpha=1.0, s=0.7, zorder=10, label=f'day {t}') # target 
            
        for t in intermediate_t:
            X1_intermediate_vis = pca.transform(mats[t])
            ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color='lightgray', 
                        alpha=0.3, s=0.7, zorder=1, label=f'day {t}')
        #ax.scatter(vis_all_days[:,0], vis_all_days[:,1], color='lightgray', 
        #           alpha=0.3, s=0.7, zorder=1) # backgrounds in gray
        #leg = ax.legend(loc='upper right')
        ttl = ax.text(0.5,1.05, "t = %.3f" % 0.0, \
                      bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                      transform=ax.transAxes, ha="center")
        ims.append([im, ttl])#, leg])
        for i, X1_trpt in enumerate(X1_trpts):  # trajectories
            if np.isnan(X1_trpt).any():
                break
            X1_hat_vis = pca.transform(X1_trpt)
            im = ax.scatter(X1_hat_vis[:,0], X1_hat_vis[:,1], color=colors[days[0]], 
                            alpha=1.0, s=0.7, zorder=10, label=f'day {days[0]}') # transported source 
            for t in days[1:]:
                X2_vis = pca.transform(mats[t])
                ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[t], 
                   alpha=1.0, s=0.7, zorder=10, label=f'day {t}') # target 
            
            for t in intermediate_t:
                X1_intermediate_vis = pca.transform(mats[t])
                ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color='lightgray', 
                        alpha=0.3, s=0.7, zorder=1, label=f'day {t}')
            
            #leg = ax.legend(loc='upper right')
            ttl = ax.text(0.5,1.05, "t = %.3f" % pts[i], \
                          bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                          transform=ax.transAxes, ha="center")
            ims.append([im, ttl])#, leg])
        ani = animation.ArtistAnimation(fig, ims, interval=50, blit=True, repeat_delay=200)
        writergif = animation.PillowWriter(fps=3)
        ani.save(img_src, writer=writergif)
        plt.clf()
        display(Image(filename = img_src))
        


    # compare W2 error with intermediate timepoints
    img_src2 = result_dir + result_name + '-w2distances.png' 
    if os.path.exists(img_src2): # try loading saved movie
        display(Image(filename = img_src2))
    else:
        W2_s = []
        if type(target_t) == type([]):
            all_ts = np.concatenate(([source_t], intermediate_t, target_t))
        else:
            all_ts = np.concatenate(([source_t], intermediate_t, [target_t]))
        all_ts = np.sort(all_ts)

        for i, X1_trpt in enumerate(X1_trpts): # trajectories
            W2_s.append([W2(X1_trpt, mats[day]).item() for day in all_ts])
            #print(f"{ts[i]}: {W2_s[j][i]}")

        plt.plot(range(0, epochs, param['save_iter']), W2_s)
        plt.legend(["Day" + str(x) for x in all_ts])

        argmins = np.argmin(np.array(W2_s), axis=0) * param['save_iter']
        print("minimum values are obtained at x=", argmins)

        plt.title(r'$W_2(P_n, Q_{day})$')  
        plt.savefig(img_src2)
        plt.show()
    
        
````

* ```source_t``` ; integer scalar: source day
* ```target_t``` ; integer list: list of target days that are informed to GPA
* ```intermediate_t```; integer list: list of intermediate days that are not informed to GPA but is wanted to be verified
* ```iter_transient_target```; integer list: list of iteration counts to reach intermediate targets
* ```total_iter```; integer scalar: iteration count to reach the target
* ```generative_model```; string: the script name in ```scripts/GPA_NN``` directory depending on nongradient drift type 'GPA_NN_w_simple_nongradient_drift'
* ```ratio_nongrad_2_grad```; float scalar: ratio of nongradient vectorfield to gradient vectorfield
* ```exp_memo```; string: naming on the experiment run

## Time-dependent vectorfield

* Step 1: For each subinterval of time, run original GPA from the cell above
* Step 2: Train time-dependent vectorfield by running scripts/train_time_dep_vectorfields.py <br>
  e.g. ```python3 train_time_dep_vectorfields.py --dataset Test_transport_genes --ts 0 2 4 --files KL-Lipschitz_1.0000-0_2times_0239_0239_00_exp1.pickle KL-Lipschitz_1.0000-2_4times_0239_0239_00_exp1.pickle --hidden_units 128 128 128 128 --mini_batch_size_t 25 --exp_memo KE=5e-3-t_size=25-network=128_128_128_128 --threshold_type KE --threshold_val 0.005 --lr_NN 0.00001 --iterations 20000```
* Step 3: Evaluate sample trajectory by running below.

````python
import tensorflow as tf
result_dir = '%s/assets/Test_transport_genes/' % main_dir

def load_velocityfield(filename):
    velocityfield = tf.keras.models.load_model(filename)

    with open(filename + "/hyperparameters.pickle", "rb") as f:
        hyperparam = pk.load(f)

    velocityfield.build(input_shape=(None, hyperparam['space_dim']+1))

    return velocityfield, hyperparam

def time_integration(x0, T, dt):
    x = tf.constant(x0, dtype=tf.float32)

    xs = [x0]
    for i in range(int(T/dt)):
        xt = tf.concat([x, dt*i*tf.ones([x.shape[0], 1], dtype=tf.float32)], axis=1)
        vv = velocityfield(xt)
        x += dt * vv
        xs.append(x.numpy())
    return xs


def generate_animation(days, intermediate_days, X1_trpts, dt, physical_dt, img_src, vs = None):
    fig, ax = plt.subplots()
    ims = []

    for i, X1_trpt in enumerate(X1_trpts):  # trajectories
        if np.isnan(X1_trpt).any():
            break
        X1_trpt_vis = pca.transform(X1_trpt)
                
        if type(vs) != type(None) and i < len(X1_trpts)-1:
            X1_trpt_vis_next = pca.transform(X1_trpts[i+1])
            vs_vis = (X1_trpt_vis_next-X1_trpt_vis) / dt
            im = ax.quiver(X1_trpt_vis[:,0], X1_trpt_vis[:,1], vs_vis[:,0], vs_vis[:,1])
        else:
            im = ax.scatter(X1_trpt_vis[:,0], X1_trpt_vis[:,1], color=colors[days[0]], 
                        alpha=1.0, s=0.7, zorder=20, label=f'day {days[0]}') # transported source  
        
            for t in days[1:]:
                X2_vis = pca.transform(mats[t])
                ax.scatter(X2_vis[:,0], X2_vis[:,1], color=colors[t], 
                   alpha=1.0, s=0.7, zorder=10, label=f'day {t}') # target 
            
            for t in intermediate_days:
                X1_intermediate_vis = pca.transform(mats[t])
                ax.scatter(X1_intermediate_vis[:,0], X1_intermediate_vis[:,1], color='lightgray', 
                        alpha=0.3, s=0.7, zorder=1, label=f'day {t}')
        
        #leg = ax.legend(loc='upper right')
        ttl = ax.text(0.5,1.05, "t = %.3f" % (physical_dt*i), \
                      bbox={'facecolor':'w', 'alpha':0.5, 'pad':5}, \
                      transform=ax.transAxes, ha="center")
        ims.append([im, ttl])#, leg])
        
    ani = animation.ArtistAnimation(fig, ims, interval=50, blit=True, repeat_delay=200)
    writergif = animation.PillowWriter(fps=3)
    ani.save(img_src, writer=writergif)
    plt.clf()
    display(Image(filename = img_src))


def generate_W2distance_plot(days, intermediate_days, X1_trpts, physical_dt, img_src):
    # compare W2 error with intermediate timepoints
    W2_s = []
    all_ts = np.sort(days + intermediate_days)

    for i, X1_trpt in enumerate(X1_trpts): # trajectories
        W2_s.append([W2(X1_trpt, mats[day]).item() for day in all_ts])

    
    plt.plot(np.arange(0, len(X1_trpts)) * physical_dt, W2_s)
    plt.legend(["Day" + str(x) for x in all_ts])

    argmins = np.argmin(np.array(W2_s), axis=0) * physical_dt
    print("minimum values are obtained at x=", argmins)

    plt.title(r'$W_2(P_n, Q_{day})$')  
    plt.savefig(img_src)
    plt.show()
    
````

````python
exp_name = 'KE=5e-3-t_size=25-network=128_128_128_128_1'
filename = result_dir + exp_name + ".pickle"
velocityfield, p = load_velocityfield(filename)

dt = p['numerical_ts'][-1]/200
X1_trpts = time_integration(mats[0], T = p['numerical_ts'][-1], dt = dt)

physical_dt = dt * p['ts'][-1] / p['numerical_ts'][-1]

img_src = result_dir + exp_name + '-movie-original.gif' 
img_src1 = result_dir + exp_name + '-movie-original-with-arrows.gif'
if os.path.exists(img_src): # try loading saved movie
    display(Image(filename = img_src))
else:
    generate_animation([0,2,4], [1,3], X1_trpts, dt, physical_dt, img_src)
if os.path.exists(img_src1): # try loading saved movie
    display(Image(filename = img_src1))
else:
    generate_animation([0,2,4], [1,3], X1_trpts, dt, physical_dt, img_src1, vs = "vecotorfield")

img_src2 = result_dir + exp_name + '-w2distances.png'
if os.path.exists(img_src2): # try loading saved image
    display(Image(filename = img_src2))
else:
    generate_W2distance_plot([0,2,4], [1,3], X1_trpts, physical_dt, img_src2)
````
