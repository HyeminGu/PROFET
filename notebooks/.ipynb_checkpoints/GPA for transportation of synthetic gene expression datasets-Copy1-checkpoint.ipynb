# GPA for transportation of synthetic gene expression datasets


This is the baseline model for transporting time series of gene expression data.

For each time points, set of gene expression samples form empirical distributions at time $t$, and say $Q_t$.

GPA algorithm implements transportation of one "empirical" measure to another "empirical" measure by calculating their $f$-divergence (metric among measures) 
$$D_f(P\|Q) = E_Q\left[f\left(\frac{dP}{dQ}\right)\right], P \ll Q$$ and form vector fields $v(x,t)$ that is the gradient descent direction of the $f$-divergence. The evolution of the measure $P$ satisfies the gradient flow
$$\partial_t P + \nabla \cdot \left(P v\right) = \partial_t P - \nabla \cdot \left(P \nabla f'\left(\frac{dP}{dQ}\right) \right) = 0$$
in the weak sense.
Here are some examples of $f$-divergences:
* KL divergence $D_\text{KL}(P\|Q) = E_P\left[\log \frac{dP}{dQ}\right] = $ with $f(x)=xlogx$
* Reverse KL divergence $D_\text{KL}(Q\|P) = E_Q\left[\log \frac{dQ}{dP}\right]$ with $f(x)=-logx$
* $\alpha$-divergence $D_\alpha(P\|Q)$ with $f(x)=\frac{x^\alpha-1}{\alpha (\alpha-1)}$ for $\alpha > 1$
* Jensen-Shannon divergence with $f(x) = -(x+1) \log \frac{1+x}{2} + x \log x$

Denote $\hat{f}$ be the Legendre transform of $f$. The variational formulation of $f$ divergences $$D_f(P\|Q)=\sup_{\phi \in C_b} E_p[\phi]-E_Q[\hat{f}(\phi)]$$
results in the PDE
$$\partial_t P - \nabla \cdot \left(P \nabla \phi\right) = 0.$$
For the stability of the algorithm, we use Lipschitz regularized function optimizer $\phi^*$ that optimize the dual formulation of $f$-divergence. 

In GPA based particle method, the vectorfield to transport particles is given as  $\dot{x} = -\nabla \phi^*(x, t)$ where $\phi^*(x, t)$ is the nonlinear optimizer $\phi^*$ above learned by Machine learning and $t$ is time for GPA algorithm.

## Aim for this notebook

Set $Q_{t_1}$ as a source distribution and $Q_{t_2}$ as a target distribution for $t_1 < t_2$. <br>In this approach, the algorithm does not know the intermediate distribution $Q_{t'}$ for $t_1 < t' < t_2$.

> <u>Future direction</u>: Constrain the trajectory with known intermediate transient distributions. <br>When $s$ is a real time (e.g. days) and the GPA algorithm time $t$ and $s$ are related by $t=ks$ for some $k>0$, assume that target distributions $Q_s$ at $s=s_1, s_2, \cdots, s_n$ are given. <br> For the smallest known index $i$ and  $t < s_i k$, we use constrained target distributions $\tilde{Q}_i=C \sum_{j=i}^n e^{-s_j k} Q_{s_j}$ where $C$ is the normalizing constant. In the particle method, it is interpreted as using entire data but reweighting individual subsets by the time order. The weights (importance) of subsets decay exponentially in real time $s$.

## Transport source to target using GPA
* Put ```--dataset Transport_genes -N_dim (d_red) --label (day1) (day2) --reduction (dim_red_method) --exp_no (exp_memo)``` and additional GPA training parameters that you want to temporarily change for this experiment. 
* Default GPA parameters and hyper-parameters are listed in ```configs/Transport_genes-GPA_NN.yaml```.
* Recommended tunable parameters depending on examples are ```--epochs``` for number of transportation steps, ```--lr_P``` for time step size for the ODE, ```--epochs_phi``` for number of gradient ascent updates for $\phi$, ```--lr_phi``` gradient ascent stepsize for $\phi$.
* For the loss function, 
    * the default setting is the Lipschitz regularized KL divergence in Donsker-Varadhan (DV) formulation; ```--f KL --formulation DV -L 1.0``` 
    * But if KL divergence is not suitable for the loss, we can try another Lipschitz regularized $f$-divergence in Legendre transform (LT) formulation; <br> ex) ```--f alpha -alpha 2.0 --formulation LT -L 1.0```
* Lipschitz constant $L$ is an algorithm level parameter. ```-L 1.0``` makes the algorithm stable in general. 
* `--` tag means required parameter, `-` tag means optional parameter defined in ```scripts/util/input_args.py```.
* The results will be saved in ```assets/Transport_genes/```. Results include 
    * movie for the evolving particles in the designated 2 axes ```plot_axes``` and ```plot_axes_join```
    * movie for the evolving vector field in the designated 2 axes ```plot_axes``` and ```plot_axes_join```
    * trajectories of particles in time in pickle format.

Steps for transportation
1. transport between consecutive time points
2. transport between not adjacent time points for prediction
3. transport using GPA with designed potential 
    - potential consits of nonlinear part (-> nonlinear residual dynamids) + quadratic part (-> linear dynamics)
    $$\phi(x, t) = \phi_{ML}(x, t) + \frac{1}{2} x^T A_t^TA_t x + b_t x$$
    - caution: Lipschitz regularized GPA would prefer quadratic terms and therefore some mollification/proximals is needed 
    - Assuming $A_t$ sparse, quadratic potential would not be critical

    - $A_t$ is given as prior knowledge which is constant in time, then we can update $A$ and also find how $A_t$ evolves in time 
    - needs validation
    
(4. traverse multiple time points -> beyond GPA)

````python
def calculate_time_steps(dt, iter_nos, physical_time=True):
# calculate real_time / iteration count
    if physical_time == True:
        iter_nos_new = []
        if type(dt) == list: # decaying or varying dt
            for iter_no in iter_nos:
                if iter_no == 0:
                    iter_nos_new.append(0)
                else:
                    iter_nos_new.append(sum(dt[:iter_no]))
        else: # constant dt
            for iter_no in iter_nos:
                iter_nos_new.append(dt*iter_no)
        iter_nos = iter_nos_new
    return iter_nos
````

### Performance metric
1. $D_f^{\Gamma_L}$ divergence : used for GPA <br>Divergence of reconstructed data and original data in the original space is bounded by the bounded by the same divergence in the latent space. (In PCA, the decoder is isometry. Therefore a metric calculated in the latent space is the same as the metric calculated in the reconstructed space.) <br> Note that the divergence is not symmetric on the order of datasets.
2. $W_2$ metric / sinkhorn divergence: objective metric <br>It is a metric!

````python
# Load f_Lip divergence calculating function
from scripts.GPA_NN.loss_NN import f_Lip_divergence
from functools import partial

f_Lip_div = partial(f_Lip_divergence, params={'f':f, 'L':1.0, 'alpha':alpha})
````

````python
# Load W2 calculating function
try:
    from geomloss import SamplesLoss
except:
    !pip3 install geomloss
    from geomloss import SamplesLoss
import torch
def sinkhorn(X, Y):
    X = torch.from_numpy(X).type(torch.float32)
    Y = torch.from_numpy(Y).type(torch.float32)

    return SamplesLoss(loss='sinkhorn', p=2)(X, Y).numpy()
````
