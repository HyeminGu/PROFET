# GPA for transportation of gene expression datasets


This is the baseline model for transporting time series of gene expression data.

For each time points, set of gene expression samples form empirical distributions at time $t$, and say $Q_t$.

GPA algorithm implements transportation of one "empirical" distribution to another "empirical" distribution by calculating their $f$-divergence (metric among distributions) and form vector fields that is the gradient descent direction of the $f$-divergence.

For the stability of the algorithm, we use Lipschitz regularized function optimizer $\phi^*$ that optimize the dual formulation of $f$-divergence.

Set $Q_{t_1}$ as a source distribution and $Q_{t_2}$ as a target distribution for $t_1 < t_2$.

The vectorfield to transport particles is given as  $\dot{x} = -\nabla \phi^*(x, t)$ where $\phi^*(x, t)$ is the nonlinear optimizer $\phi^*$ above learned by Machine learning and $t$ is time for GPA algorithm.

### Generate source and target datasets for GPA
* For now, we will apply GPA in order to transport one empirical distribution (source) to the next adjacent day's empirical distribution (target).
* In this step, we prepare source and target datasets that can be fed as GPA input.
* GPA input datasets will be stored in ```data/(example name)/``` directory.
* Here, we use PCA reduced data matrices ```mats_reduced``` as original datasets. EMT Hallmark reduced data matrices ```mats_reduced2``` is another option.
* First we apply PCA to (source+target) combined data matrix.
* We save ```d_red```(=##) dimensional PCA basis in ```pca_##.pkl``` in the Pickle data format.
* The source matrix projected to ```d_red``` dimensional PCA basis is saved in ```source_dataset_dim_##.csv```.
* The target matrix projected to ```d_red``` dimensional PCA basis is saved in ```target_dataset_dim_##.csv```.

````python
from sklearn import decomposition
````

````python
d_red = 64    # reduced dimension
c1, c2 = 0, 1 # source day, target day

example_name = "Day01"

X = np.concatenate((mats_reduced[c1], mats_reduced[c2])) # source + target combined matrix
print(X.shape)
````

````python
# objective metric on two empirical distributions
from geomloss import SamplesLoss
import torch
def sinkhorn(X, Y):
    X = torch.from_numpy(X).type(torch.float32)
    Y = torch.from_numpy(Y).type(torch.float32)

    return SamplesLoss(loss='sinkhorn', p=2)(X, Y).numpy()
````

````python
# PCA on X
pca = decomposition.PCA(n_components=min(X.shape))
pca.fit(X)

variances = np.array(pca.explained_variance_)/np.sum(pca.explained_variance_)*100
#print(np.cumsum(variances))
plt.plot(range(1, min(X.shape)+1), np.cumsum(variances), '-')
plt.xlabel("number of PCA basis", fontsize=16)
plt.ylabel("Explained variance (%)", fontsize=16)
plt.vlines(x=d_red, ymin=variances[0], ymax=100, color = 'r', linestyles='dashed')
plt.text(x=d_red+7, y= np.cumsum(variances)[d_red-1]-3, s="$d'=%d$\n%.2f %%" % (d_red, np.cumsum(variances)[d_red-1]), fontsize=16)
plt.ylim([variances[0], 100])
plt.tick_params(axis='x', labelsize=14)
plt.tick_params(axis='y', labelsize=14)
plt.show()

# --------
````

````python
# PCA for d_red dimensional basis and projection
pca = decomposition.PCA(n_components=d_red, random_state=0)
pca.fit(X)

X1_red = pca.transform(mats_reduced[c1])
X2_red = pca.transform(mats_reduced[c2])

# reconstruction
X1_hat = pca.inverse_transform(X1_red)
X2_hat = pca.inverse_transform(X2_red)


print("Sinkhorn divergence on original source/target and reconstructed source/target")
print("Source: ", sinkhorn(X1_hat, mats_reduced[c1]))
print("Target: ", sinkhorn(X2_hat, mats_reduced[c2]))
````

````python
# Save PCA results
folder_name = main_dir + "/data/%s/" % example_name
if not os.path.exists(folder_name):
    os.mkdir(folder_name)
pk.dump([pca,], open(folder_name+"pca_"+str(d_red)+".pkl","wb"))

np.savetxt(folder_name+"source_dataset_dim_"+str(d_red)+".csv", X1_red, delimiter=",")
np.savetxt(folder_name+"target_dataset_dim_"+str(d_red)+".csv", X2_red, delimiter=",")
````

## Transport source to target using GPA
* Put ```--dataset (example_name)```, ```-N_dim (d_red)```, ```--exp_no (experiment memo in string)``` and additional parameters that you want to temporarily change for this experiment.
* Recomended tunable parameters depending on examples are ```--epochs``` for number of transportation steps, ```--lr_P``` for time step size for the ODE, ```--epochs_phi``` for number of gradient ascent updates for $\phi$, ```--lr_phi``` gradient ascent stepsize for $\phi$.
* For the loss function,
    * the default setting is the Lipschitz regularized KL divergence in Donsker-Varadhan (DV) formulation; ```--f KL --formulation DV -L 1.0```
    * But if KL divergence is not suitable for the loss, we can try another Lipschitz regularized $f$-divergence in Legendre transform (LT) formulation; <br> ex) ```--f alpha -alpha 2.0 --formulation LT -L 1.0```
* Lipschitz constant $L$ is an algorithm level parameter. ```-L 1.0``` makes the algorithm stable in general.
* Generic GPA parameters and hyper-parameters are listed in ```configs/Transport_genes-GPA_NN.yaml```.
* `--` tag means required parameter, `-` tag means optional parameter defined in ```scripts/util/input_args.py```.
* The results will be saved in ```assets/(example_name)/```. Results include
    * movie for the moving particles in the first 2 axis
    * vector field (to be included)
    * trajectories of particles in time in pickle format.

### Load time trajectories of particles
Results are stored in ```assets/(example name)/``` directory as pickle format file.

Replace the ```result_filename``` according to the adjusted settings from GPA.

````python
def calculate_time_steps(dt, iter_nos, physical_time=True):
# calculate real_time / iteration count
    if physical_time == True:
        iter_nos_new = []
        if type(dt) == list: # decaying or varying dt
            for iter_no in iter_nos:
                if iter_no == 0:
                    iter_nos_new.append(0)
                else:
                    iter_nos_new.append(sum(dt[:iter_no]))
        else: # constant dt
            for iter_no in iter_nos:
                iter_nos_new.append(dt*iter_no)
        iter_nos = iter_nos_new
    return iter_nos
````

````python
trans_red_dset1_name = main_dir + '/assets/' + example_name + '/' + result_filename

with open(trans_red_dset1_name, "rb") as fr:
    param, result = pk.load(fr)
    X1_red_trpts = result['trajectories']


ts = [int(r*param['epochs']) for r in [0.09, 0.12, 0.99999]]
print("transport time step: ", ts)
pts = calculate_time_steps(param['lr_P'], ts)
print("transport physical time step: ", pts)

X1_red_trpts = X1_red_trpts[-1]
````

### Reconstruct the particles to the original space
In case of PCA, do PCA inverse transform to reconstruct data

````python
X1_trpts = pca.inverse_transform(X1_red_trpts)
#print(X1_trpts[:10, :10])

# reconstructed transported vs original source
print("Sinkhorn divergence between reconstructed transported vs original source: ", sinkhorn(X1_trpts, mats_reduced[c1]))
# reconstructed transported vs reconstructed source
print("Sinkhorn divergence between reconstructed transported vs reconstructed source: ", sinkhorn(X1_trpts, X1_hat))
# reconstructed transported vs original target
print("Sinkhorn divergence between reconstructed transported vs original target: ", sinkhorn(X1_trpts, mats_reduced[c2]))
# reconstructed transported vs reconstructed target
print("Sinkhorn divergence between reconstructed transported vs reconstructed target: ", sinkhorn(X1_trpts, X2_hat))
````
