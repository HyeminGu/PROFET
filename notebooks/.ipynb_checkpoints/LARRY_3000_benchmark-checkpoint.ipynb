```python
import os
import sys
import pickle
import shutil
import subprocess
import numpy as np
from collections import Counter
from IPython.display import Image, display

sys.path.insert(0, os.path.abspath('../../'))
from util.utils import ResourceMonitor

main_dir = os.getcwd()
if "notebooks" in main_dir:
    main_dir = main_dir[:-10]
os.chdir(main_dir)
print("main_dir:", main_dir)

contrast_colors = {
    2: '#1f77b4',   # blue
    4: '#ff7f0e',   # orange
    6: '#d62728',   # red
}
```

## LARRY_3000_benchmark dataset

### Data property
* 3 snapshots at day 2, 4, 6
* Training days: [2, 6]; intermediate (held-out) day: [4]
* dimension: reduced to d_red=2 via PCA

### Data source
Preprocessed pickle is loaded from the shared benchmarks data directory `../../data/LARRY_3000_benchmark_preprocessed.pkl`
and copied to `data/` (source_code_v3) so GPA can locate it.

```python
# -------------------------------------------------------
# Load preprocessed data from shared benchmarks data dir
# -------------------------------------------------------
example_name = 'LARRY_3000_benchmark'

shared_pkl = os.path.abspath(os.path.join(main_dir, '../../data', f'{example_name}_preprocessed.pkl'))
local_pkl  = os.path.join(main_dir, 'data', f'{example_name}_preprocessed.pkl')

# Copy to source_code_v3/data/ so GPA scripts can find it
if not os.path.exists(local_pkl):
    shutil.copy(shared_pkl, local_pkl)
    print(f'Copied {shared_pkl} -> {local_pkl}')
else:
    print(f'Local pkl already exists: {local_pkl}')

with open(local_pkl, 'rb') as fr:
    time_label, full_matrix, projected_matrix, pca = pickle.load(fr)

time_label     = np.array(time_label, dtype=np.float32)
projected_matrix = np.array(projected_matrix, dtype=np.float32)

time_points = sorted(set(time_label))
cnt = Counter(time_label)
print('Time points:', time_points)
print('Cell counts per time point:', dict(sorted(cnt.items())))
print('full_matrix shape:', full_matrix.shape)
print('projected_matrix shape:', projected_matrix.shape)
```

## PROFET

```python
# -------------------------------------------------------
# User parameters
# -------------------------------------------------------
d_red             = 2
times             = [2, 6]       # training snapshots
intermediate_times = [4]         # held-out intermediate snapshot
dimension_reduction = 'Y'
exp_no            = 'benchmark'
exp_memo          = f'dim{d_red}_{exp_no}'

# GPA tunable parameters
adjust_lr_P_epochs    = 'Y'      # 'Y' to auto-adjust lr_P and epochs w.r.t. dimensionality
specify_f_Lip_threshold = 'Y'
f_Lip_threshold       = 1e-4

# ForceMatching tunable parameters
fm_iterations = 20000
fm_hidden_units = '64 64 64 64'
```

```python
# -------------------------------------------------------
# PROFET Step 1: GPA  (one pair: day 2 -> day 6)
# -------------------------------------------------------
os.chdir(os.path.join(main_dir, 'scripts', 'GPA'))

dimension_note = f'dim{d_red}-' if dimension_reduction in ['Y', 'yes'] else ''

gpa_filenames = []
with ResourceMonitor() as monitor:
    for i in range(len(times) - 1):
        cmd = (
            f'python3 train.py'
            f' --dataset {example_name}'
            f' --label {times[i]} {times[i+1]}'
            f' --N_dim {d_red}'
            f' --exp_no {exp_no}'
            f' --dimension_reduction {dimension_reduction}'
        )
        if adjust_lr_P_epochs in ['N', 'no']:
            cmd += f' --adjust_lr_P_epochs {adjust_lr_P_epochs}'
        if specify_f_Lip_threshold in ['Y', 'yes']:
            cmd += f' --f_Lip_threshold {f_Lip_threshold}'

        print(f'Running GPA: day {times[i]} -> day {times[i+1]}')
        subprocess.run(cmd, shell=True, check=True)

        N_Q = int(cnt[times[i+1]])
        N_P = int(cnt[times[i]])
        fname = f'KL-Lipschitz_1.0000-{times[i]}_{times[i+1]}times-{dimension_note}{N_Q:04d}_{N_P:04d}-00-{exp_no}.pickle'
        gpa_filenames.append(fname)

os.chdir(main_dir)

result_dir = os.path.join(main_dir, 'assets', example_name, exp_memo)
os.makedirs(result_dir, exist_ok=True)
monitor.report('PROFET GPA', dataset=example_name, d_red=d_red,
               filename=os.path.join(result_dir, 'resources.txt'))

print('GPA output files:', gpa_filenames)
```

```python
# -------------------------------------------------------
# PROFET Step 2: Force Matching
# -------------------------------------------------------
os.chdir(os.path.join(main_dir, 'scripts', 'ForceMatching'))

times_join         = ' '.join(str(t) for t in times)
gpa_filenames_join = ' '.join(gpa_filenames)

cmd = (
    f'python3 train.py'
    f' --dataset {example_name}'
    f' --ts {times_join}'
    f' --exp_memo {exp_memo}'
    f' --files {gpa_filenames_join}'
    f' --iterations {fm_iterations}'
    f' --hidden_units {fm_hidden_units}'
)

with ResourceMonitor() as monitor:
    print('Running Force Matching...')
    subprocess.run(cmd, shell=True, check=True)

os.chdir(main_dir)
monitor.report('PROFET ForceMatching', dataset=example_name, d_red=d_red,
               filename=os.path.join(result_dir, 'resources.txt'))
```

```python
# -------------------------------------------------------
# Load velocity field and integrate ODE
# -------------------------------------------------------
from scripts.ForceMatching.evaluations import load_velocityfield, time_integration

velocity_net, p = load_velocityfield(f'assets/{example_name}/{exp_memo}/')

dt          = p['numerical_ts'][-1] / 200
numerical_dt = dt
physical_dt  = dt * p['ts'][-1] / p['numerical_ts'][-1]

input_data = projected_matrix[time_label == times[0], :d_red]

X1_trpts = time_integration(input_data, T=p['numerical_ts'][-1], dt=dt, velocity_net=velocity_net)
print(f'Trajectory frames: {len(X1_trpts)}, shape per frame: {X1_trpts[0].shape}')

# Save trajectories
traj_npy = os.path.join(result_dir, 'backward_trajectories.npy')
np.save(traj_npy, np.stack(X1_trpts, axis=0))
print(f'Saved trajectory to {traj_npy}')
```

```python
# -------------------------------------------------------
# Visualization
# -------------------------------------------------------
from scripts.util.visualizations import animate_trajectories, plot_w2errors

img_src1 = os.path.join(result_dir, 'movie-particles.gif')
img_src2 = os.path.join(result_dir, 'movie-velocities.gif')
img_src3 = os.path.join(result_dir, 'w2distances.png')

animate_trajectories(
    X1_trpts, times, intermediate_times,
    time_label, projected_matrix,
    img_src1, numerical_dt, physical_dt=physical_dt,
    colors=contrast_colors, dimension_reduction=True, plot_vectorfield=False,
)
animate_trajectories(
    X1_trpts, times, intermediate_times,
    time_label, projected_matrix,
    img_src2, numerical_dt, physical_dt=physical_dt,
    colors=contrast_colors, dimension_reduction=True, plot_vectorfield=True,
)

reference_matrix = projected_matrix[:, :d_red]
all_times = sorted(times + intermediate_times)
plot_w2errors(
    X1_trpts, time_label, all_times,
    img_src3, reference_matrix, physical_dt,
    colors=contrast_colors,
)
```
