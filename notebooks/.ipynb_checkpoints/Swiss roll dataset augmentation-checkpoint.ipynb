```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pickle
import matplotlib as mpl
from mpl_toolkits.axes_grid1.inset_locator import zoomed_inset_axes
from mpl_toolkits.axes_grid1.inset_locator import mark_inset
```

```python
label_names = ['200 original samples', '200+5000 augmented samples']
sample_sizes = [200, 5200]
common_names = '../assets/Swiss_roll/assets/'
colors = ['r', 'b']
last_trajectories = []
targets = []

fig, ax = plt.subplots()
# zoomed part
xlims = [18500, 20000]
ylims = [0.1, 1.0]
x = range(1, 20002)
x_start = 1
axins = zoomed_inset_axes(ax, 4, loc='center right', )#axes_kwargs={"facecolor": "lightgray"})

for label_name, sample_size, color in zip(label_names, sample_sizes, colors):
    file_name = common_names+'samplesize_%d_iter_20000.pickle' % sample_size
    with open(file_name, "rb") as fr:
        d_losses, trajectories, samples, methods, GP_type, learning_rates, optimizer = pickle.load(fr)
        
    last_trajectories.append(trajectories[-1][-1])
    targets.append(samples)
    
    ax.plot(x[x_start:], d_losses[-1][x_start:], label = label_name, c=color)
    
for label_name, sample_size, color in zip(label_names, sample_sizes, colors):
    file_name = common_names+'samplesize_%d_iter_20000.pickle' % sample_size
    with open(file_name, "rb") as fr:
        d_losses, trajectories, samples, methods, GP_type, learning_rates, optimizer = pickle.load(fr)
        
    last_trajectories.append(trajectories[-1][-1])
    targets.append(samples)
    # zoomed part
    axins.plot(x[x_start:], d_losses[-1][x_start:], label = label_name, c=color)

axins.set_xlim(xlims)
axins.set_ylim(ylims)
axins.grid('on')

ax.set_xlim([x[x_start], x[-1]])
ax.set_ylim([0, 10])
mark_inset(ax, axins, loc1=1, loc2=3, fc='lightgray', ec='0.5')
ax.grid('on')
ax.legend(fontsize="16", frameon=False)
ax.set_xlabel('iterations')
ax.set_ylabel('KL-Lip Divergence')

plt.draw()
plt.show()

    
```

```python
for target, last_trajectory, color in zip(targets, last_trajectories, colors):
    plt.scatter(target[:,0], target[:,2], zorder=1, label='Training samples', s=5, alpha=1.0, c='limegreen')
    plt.scatter(last_trajectory[:,0], last_trajectory[:,2], zorder=10, label='Generated samples',s=5, alpha=0.3, c=color)
    plt.legend(fontsize="16")
    plt.xlim([-18, 18])
    plt.ylim([-18, 27])
    plt.show()
```
